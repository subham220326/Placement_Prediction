```python
#now we will distingish input and output 
x=df.iloc[:,0:2]
y=df.iloc[:,2:3]
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test,y_train,y_test=train_test_split(x,y,test_size=0.1)#this function helps 
#to distigusih the test subject and train subject and here we are using 10% of data for testing
#and remaning data for training the model
```

```python
X_train
```

```python
y_train
```

```python
X_test
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()#StandardScaler is used to standardize the data by removing 
#the mean and scaling to unit variance (-1 to 1)
```

```python
X_train=scaler.fit_transform(X_train)#here the function in it transforms the values form -1 to 1
#just google what is fit and transform in sklearn.
```

```python
X_train
```

```python
X_test=scaler.transform(X_test)
#here we are transforming the test data to the same scale as the train data
```

```python
X_test
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf=LogisticRegression()#here we are using the logistic regression model to train the data
```

```python
clf.fit(X_train,y_train)
```

```python
clf.predict(X_test)#here we are predicting the output of the test data using the trained model
```

```python
y_test#here we are checking the actual output of the test data to compare with the predicted output
```

```python
y_pred=clf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
#here we are checking the accuracy of the model
#  by comparing the actual output and predicted output
```

```python
from mlxtend.plotting import plot_decision_regions
#this function is used to plot the decision regions of the model
```

```python
plot_decision_regions(X_train, y_train.values.ravel(), clf=clf, legend=2)
```
